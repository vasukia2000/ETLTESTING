# Filter ( where caluse in sql)

```python
import pandas as pd
df = pd.read_csv("emp_details.csv")
```

```python
# Filter ( where clause in SQL)
# Q1 . Get me all the employees for deptno = 10
# SQL  select * from emp where deptno=10
deptno_10 = df['deptno'] == 10
df[deptno_10]
```

```python
# Filter ( where clause in SQL)
# Q1 . Get me all the employees ther than  deptno = 10
# SQL  select * from emp where deptno!=10
deptno_10 = df['deptno'] == 10
df[~deptno_10]
```

```python
# Filter ( where clause in SQL)
# Q1 . Get me all the employees for deptno = 10 or 20
# SQL  select * from emp where deptno in (10,20)
deptno_10 = df['deptno'] == 10
deptno_20 = df['deptno'] == 20
deptno_10_and_20 = deptno_10 | deptno_20
df[deptno_10_and_20]
```

```python
# Filter ( where clause in SQL)
# Q1 . Get me all the employees for deptno = 10 whose salaey is less than or equal 10,00
# SQL  select * from emp where deptno= 10 and salary <=10000
deptno_10 = df['deptno'] == 10
salary = df['salary'] <= 1000
final_filter = deptno_10 & salary
df[final_filter]
```

```python
# Filter ( where clause in SQL)
# Q1 . Get me all the employees for deptno = 10 whose salary not equals equal to 10,00
# SQL  select * from emp where deptno= 10 and salary !=1000
deptno_10 = df['deptno'] == 10
salary = df['salary'] == 1000
salary = ~salary
final_filter = deptno_10 & salary
df[final_filter]
```

```python
# Sorting can be done in two diffrent ways : sort_values or sort_index
# Q1 : sort the employess rows in descending order of salary ( higest to lowest salary)
# SQL : select * from emp order by salary desc
df.sort_values(by='salary',ascending=False)
```

```python
# Q1 . Get me the count of employees, total_salary,max_salary,min_salary per department
# SQL : select deptno,count(*),sum(sal),max(sal),min(sal) from emp group by deptno
grped_df = df.groupby('deptno')['salary'].agg(['sum','max','min','count']).reset_index()

grped_df.rename(columns = {"sum":"Total_salary","max":"Maximun_salary","min":"Minimum_salary","count":"employees count"},inplace=True)
```

# merging ( SQL join equivalent)

```python
df_emp = pd.read_csv("emp.csv")
df_sal = pd.read_csv("salary.txt")
```

```python
df_sal
```

```python
df_emp
```

```python
# Merge Syntax ( inner join/ equi join)
# SQL : select e.eno,e.ename,s.salary from emp as e inner join salary as s on e.eno = s.eno
df_emp.merge(df_sal,on='eno',how='inner')
```

```python
# Merge Syntax ( left outer join)
# SQL : select e.eno,e.ename,s.salary from emp as e left join salary as s on e.eno = s.eno
df_emp.merge(df_sal,on='eno',how='left')
```

```python
# Merge Syntax ( right outer join)
# SQL : select e.eno,e.ename,s.salary from emp as e right join salary as s on e.eno = s.eno
df_emp.merge(df_sal,on='eno',how='right')
```

```python
# Merge Syntax ( inner join)
# SQL : select e.eid,e.ename,s.salary from emp as e inner join salary as s on e.eid = s.eno
df_emp.merge(df_sal,left_on='eid',right_on='eno',how='inner')
```
