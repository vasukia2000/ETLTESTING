```python
from sqlalchemy import create_engine
import oracledb
```

```python
mysql_engine = create_engine("mysql+pymysql://root:Admin%40123@localhost:3306/etl")
oracle_engine = create_engine("oracle+oracledb://system:Admin@localhost:1521/xe")
query_mysql = "select * from Customers"
query_oracle = "select * from customers"
```

```python
df_mysql =pd.read_sql(query_mysql,mysql_engine)
```

```python
df_mysql
```

```python
df_oracle =pd.read_sql(query_oracle,oracle_engine)
```

```python
df_oracle
```

```python
df_mysql =pd.read_sql(query_mysql,mysql_engine)
df_mysql.to_sql("customer_info",oracle_engine,index = False)
```

```python
oracle_connect = oracledb.connect(
    user = "system",
    password = "Admin",
    dsn = "localhost:1521/xe")
query_oracle = "select * from customers"
```

```python
df_oracle =pd.read_sql(query_oracle,con = oracle_connect)
```

```python
mysql_engine_scr = create_engine("mysql+pymysql://root:Admin%40123@localhost:3306/etl")
oracle_engine = create_engine("oracle+oracledb://system:Admin@localhost:1521/xe")
query_mysql_scr = "select * from Customers where CustomerID ='C001' "
query_oracle_tar = "select * from customer_info"
```

```python
df_scr = pd.read_sql(query_mysql_scr,mysql_engine_scr)
df_tar = pd.read_sql(query_oracle_tar,oracle_engine)
```

```python
df_scr
```

```python
df_tar
```

```python
if df_scr.equals(df_tar) ==False:
  print("data not matchied")
else :
  print("Data match")
```

```python
assert df_scr.equals(df_tar) ,"data not matchied"
```

```python
df_scr['CustomerID']
```

```python
df_scr['CustomerID'].isin(df_tar['CustomerID']) #one particular row is there or not
```

```python
status =  df_scr['CustomerID'].isin(df_tar['CustomerID'])
```

```python
status
```

```python
count_matched = status.sum()
```

```python
assert count_matched >1,"no match"
```
